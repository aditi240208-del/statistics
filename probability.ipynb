```python

## Discrete Probability Distributions — Binomial, Poisson & Geometric

# Import the libraries

import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
# Set the random seed - This makes the random simulation reproducible.

np.random.seed(42)
# 1. Define the Binomial Distribution
n, p = 20, 0.35
binom = stats.binom(n, p)
x = np.arange(0, n+1)
# Mean of Binomial distribution
print(f' Mean = {binom.mean():.4f} (Theory: np={n*p})')
# Variance
binom.var()
# Second Moment E[X^2]

binom.moment(2)
# Skewness

binom.stats("s")
# CDF – Probability of at most 10 successes

binom.cdf(10)
# PMF – Probability of exactly 7 successes

binom.pmf(7)
# --- VISUALIZATION ---

# 1. Binomial PMF
plt.figure(figsize=(7, 5))
plt.bar(x, binom.pmf(x), color='steelblue',
        edgecolor='black', alpha=0.8)

plt.axvline(binom.mean(), color='red', linestyle='--',
            label=f'Mean = {binom.mean():.1f}')

plt.title(f'Binomial PMF (n={n}, p={p})')
plt.xlabel('Number of Successes (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 2. Poisson Distribution

for lam in [1, 3, 5, 10]:
    d = stats.poisson(lam)
    sim = np.random.poisson(lam, 5000)
# Theoretical Mean

d.mean()
# Simulation - This generates 5,000 random observations from the Poisson distribution.

sim = np.random.poisson(lam, 5000)
# Simulation mean - calculates the average of the simulated observations.
sim.mean()
# calculation of all together

for lam in [1, 3, 5, 10]:
    d = stats.poisson(lam)
    sim = np.random.poisson(lam, 5000)
    print(f'Poisson(λ={lam}): Th.Mean={d.mean()}, Sim.Mean={sim.mean():.3f}')
# Visualization - Poisson PMF
plt.figure(figsize=(7, 5))

x_p = np.arange(0, 20)

for lam in [1, 3, 5, 10]:
    plt.plot(x_p, stats.poisson.pmf(x_p, lam),
             marker='o', markersize=4, label=f'λ = {lam}')

plt.title('Poisson PMF — Effect of λ')
plt.xlabel('Number of Events (x)')
plt.ylabel('Probability')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
# 3. GEOMETRIC DISTRIBUTION

p_geo = 0.3
geo = stats.geom(p_geo)
# Mean

geo.mean()
# Probability of success on the first trial

geo.pmf(1)
# Probability that first success takes more than 5 trials

1 - geo.cdf(5)
print(f'Geometric(p={p_geo}):')
print(f'  Mean = 1/p = {geo.mean():.4f}')
print(f'  P(X=1) = {geo.pmf(1):.4f}')
print(f'  P(X>5) = {1 - geo.cdf(5):.4f}')
# Visualization - Geometric PMF
plt.figure(figsize=(7, 5))

x_g = np.arange(1, 20)

plt.bar(x_g, geo.pmf(x_g), color='green',
        edgecolor='black', alpha=0.8)

plt.title(f'Geometric PMF (p={p_geo})')
plt.xlabel('Trial Number (x)')
plt.ylabel('Probability')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
